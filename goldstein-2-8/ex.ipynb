```python
import sympy as smp
```

```python
a, b, t, m, g = smp.symbols("a b t m g", real=True)
```

```python
y = a * t + b * t**2
y
```

```python
y0 = smp.symbols("y_0", real=True)
t0_eq = smp.sqrt(2 * y0 / g)
t0_eq
```

```python
T = 1/2 * m * smp.diff(y, t)**2
V = m * g * y
L = T - V
L
```

```python
L.subs({t: t0_eq}).expand()
```

Find the extremum of the function (maximum or minimum)

$$ \int_0^{t_0} L~dt $$

To do that we have to find the where the derivative is $0$. We already have the derivative which is the Lagrangian $L$.

To find the solutions we simply a system where we plug the Lagrangian and $y(t_0) = y_0$.

```python
system = smp.Matrix(
    [L.subs({t: t0_eq}).expand(), y.subs({t: t0_eq}) - y0],
)
system
```

```python
sols = smp.solve(
    system,
    [a,b],
)
```

```python
sols
```

By solving the linear system two set of solutions are found for $(a, b)$

```python
smp.Matrix(sols[0])
```
